## Merger 27 File data udara
#### untuk setiap file kita tambah kolom `Location` untuk nantinya berisi value LOC-[Kode Lokasi]. Dengan tujuan (ketika di merge) informasi mengenai lokasi dari setiap baris data masih ada.

#### meski jumlah dan data dari kolomnya sudah sama, dalam kolom parameter (seperti pm25, so2, co, dan sebagainya) memiliki satuan yang berbeda (ug_m3 dan ppm). kita harus `standarisasi` ini menjadi ug_m3 semua supaya bisa di merge.

```python
df = pd.concat(table_data_udara, ignore_index=True)
df.sample(5)
```

```python
# convert ppm to ug/m3

# define conversion factors for each pollutant
conversion_factors = {
    'pm25': 1000,
    'pm10': 1000,
    'so2': 2620,
    'no2': 1880,
    'co': 1145,
    'o3': 1963
}

# loop for fill ug_m3 column (ppm * conversion factor)
for param, factor in conversion_factors.items():
    col_ug = f"{param}_ug_m3"
    col_ppm = f"{param}_ppm"
    
    if col_ug in df.columns and col_ppm in df.columns:
        df[col_ug] = df[col_ug].fillna(df[col_ppm] * factor)

# hapus kolom ppm
ppm_columns = [f"{param}_ppm" for param in conversion_factors.keys()]
df = df.drop(columns=ppm_columns, errors='ignore')
```

```python
# change measurement_id

# order data (partition by location) and sort by timestamp_start
df = df.sort_values(by=['location', 'timestamp_start']).reset_index(drop=True)

# generate sequence number for each location
sequence_number = df.groupby('location').cumcount() + 1
location_clean = df['location'].astype(str).str.replace('LOC-', '', regex=False)

# manipulate measurement_id column
df['measurement_id'] = 'MEAS-' + location_clean + '-' + sequence_number.astype(str).str.zfill(5)
```
