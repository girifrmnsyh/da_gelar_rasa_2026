#### **STEP 2: STANDARIZATION VALUE COLUM**
Menstandarisasi value dari kolom:
1. `kota_kabupaten`
2. `kecamatan`<br>
pada kolom **sites** dan **hospitals** untuk value yang memiliki ketidakteraturan format, seperti:
- **Kota Administrasi Jakarta Pusat** dan **Jakarta Pusat** yang harusnya sama.
- **Cilincing (Marunda)** dan **Cilincing** yang harusnya sama.

##### standarization `kota_kabupaten` colum

```python
sites['kota_kabupaten'].unique()
```

```python
hospitals['kota_kabupaten'].unique()
```

```python
sites['kota_kabupaten'].unique() == hospitals['kota_kabupaten'].unique()
```

```python
# replace 'Kota Administrasi' to ''
sites['kota_kabupaten'] = sites['kota_kabupaten'].astype(str).str.replace('Kota Administrasi', '', regex=False).str.strip()
sites['kota_kabupaten'].unique() == hospitals['kota_kabupaten'].unique() #check
```

##### standarization `kecamatan` colum

```python
sites['kecamatan'].unique()
```

```python
hospitals['kecamatan'].unique()
```

```python
# check
sites['kecamatan'].unique() == hospitals['kecamatan'].unique()
```

```python
# standarization 'kecamatan'
sites['kecamatan'] = sites['kecamatan'].astype(str).str.replace(r'\(.*?\)', '', regex=True).str.strip()
sites['kecamatan'].unique() == hospitals['kecamatan'].unique() #check
```

```python
review(sites) # review result
```

#### **STEP 3: TABLE MODIFICATION**
- mengganti kolom `[parameter]_unit` menjadi kolom `unit` dengan value **ug/m3**. (karena kita mengganti semua unit pengukuran menjadi ug/m3 di table **aqms_data**)
- mengubah nama table menjadi bahasa inggris (gunakan 1 bahasa)

```python
# drop & add colums at sites table
sites = sites.drop(columns=['pm25_unit','pm10_unit','so2_unit',
                            'co_unit','o3_unit','no2_unit'])
sites.insert(20,'unit','ug/m3')

review(sites)
```

```python
# drop colums at sensors
sensors = sensors.drop(columns=['latitude','longitude'])
review(sensors)
```

```python
# reference
print(sensors.columns)
print(sites.columns)
print(hospitals.columns)
```

```python
# modification table colum
sensors.columns = ['sensor_id', 'location_id', 'sensor_name', 'sensor_model',
       'installation_type', 'operational_status',
       'installation_date', 'region_group']

sites.columns = ['location_id', 'province', 'city', 'district',
       'regional_characteristics', 'latitude', 'longitude', 'baseline_pm25',
       'region_group', 'station_code', 'site_type', 'distance_to_road_m',
       'traffic_intensity_per_h', 'pm25_equipment', 'pm10_equipment',
       'so2_equipment', 'co_equipment', 'o3_equipment', 'no2_equipment',
       'meteo_equipment', 'unit', 'notes']

hospitals.columns = ['hospital_id', 'hospital_name', 'city', 'district',
       'province', 'facility_class', 'bed_capacity']

print('done')
```

#### **STEP 4: EXPORT DATA**

```python
# export data
sites.to_csv('../[3] CLEAN Dataset DA Gelar Rasa 2026/sites.csv', sep=',', index=False)
sensors.to_csv('../[3] CLEAN Dataset DA Gelar Rasa 2026/sensors.csv', sep=',', index=False)
hospitals.to_csv('../[3] CLEAN Dataset DA Gelar Rasa 2026/hospitals.csv', sep=',', index=False)

print('done')
```
