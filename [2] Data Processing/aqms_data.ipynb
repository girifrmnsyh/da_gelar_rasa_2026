#### **STEP 2: MERGE DATA AQMS**
untuk setiap file/table **AQMS** kita merge untuk nantinya menjadi 1 *table* utuh dengan tujuan memudahkan proses analisis.


```python
# merge all AQMS tables
aqms_tables = [bgl01, bgr01, bgr02, bjb01, btm01, idp01, jkt01, jkt02, jkt03, jmb01, jmb02, ktb01, ktg01, ktg02, mdn01, mer01, mlg01, mrs01, pkn01, pku01, plb01, plu01, psw01, sm201, srg01, stg01, yky01]

df = pd.concat(aqms_tables, ignore_index=True)
df.sample(5, random_state=45)
```

#### **STEP 3: STANDARIZATION UNIT**
*convert* data parameter polutan dengan satuan ppm ke ug_m3 (standarisasi satuan/unit pengukuran)

dengan acuan:
- blablabla

```python
# convert ppm to ug/m3

# define conversion factors for each pollutant
conversion_factors = {
    'pm25': 1000,
    'pm10': 1000,
    'so2': 2620,
    'no2': 1880,
    'co': 1145,
    'o3': 1963
}

# loop for fill ug_m3 column (ppm * conversion factor)
for param, factor in conversion_factors.items():
    col_ug = f"{param}_ug_m3"
    col_ppm = f"{param}_ppm"
    
    if col_ug in df.columns and col_ppm in df.columns:
        df[col_ug] = df[col_ug].fillna(df[col_ppm] * factor)

# hapus kolom ppm
ppm_columns = [f"{param}_ppm" for param in conversion_factors.keys()]
df = df.drop(columns=ppm_columns, errors='ignore')
```

```python
# review data
df.sample(5, random_state=45)
```

#### **STEP 4: PRIMARY KEY STANDARIZATION**
Standarisasi `measurement_id` (*primary key*) penting untuk memastikan bahwa kolom tersebut unik dan juga rapi. 
**terlihat dalam sample diatas formatnya masih tidak selaras.**<br>
seperti: MEAS-XXXXXX, MEAS-XXX-XXXXXX, atau MEAS-[Kode Lokasi]-XXXXXX

kita ubah dengan format:
**MEAS-XXXOOOOO**, <br>XXX : Kode sensor dan OOOOO : sequence number

```python
# primary key
df = df.sort_values(by=['sensor_id', 'timestamp_start']).reset_index(drop=True)

sequence_number = df.groupby('sensor_id').cumcount() + 1
sensor_code = df['sensor_id'].astype(str).str.replace('SENS-', '', regex=False)

df['measurement_id'] = 'MEAS-' + sensor_code + sequence_number.astype(str).str.zfill(5)
```

```python
# review data
df.sample(5, random_state=45)
```

```python
# checking measurement_id is unique
df['measurement_id'].nunique() == df.shape[0]
```

#### **STEP 5: CLEANING & DOCUMENTATION RAW DATA**
OBJEKTIF:
1. Membersihkan data tanggal (hapus milisecond) untuk standarisasi.
2. Mengubah tipe data kolom `pm25_ug_m3, pm10_ug_m3, so2_ug_m3, co_ug_m3, o3_ug_m3, no2_ug_m3, temperature_c, humidity_pct, wind_speed_ms, dan wind_direction_deg` menjadi *float* untuk kebutuhan analisis.
3. mendokumentasikan data asli (bikin table baru) dari baris baris dengan value non numeric pada kolom yang disebutkan diatas.

METODE/CARA:
1. Bikin Kolom baru bernama `is_valid` dengan mapping:<br>
VALID : {VALID} <br>
NOT VALID : {SUSPECT, MISSING, CALIBRATING, WARMUP, PARSER_FAIL, FATAL_ERR, GARBAGE, COMM_LOST, CORRUPT, SENSOR_DEAD, dan MAINTENANCE}
2. Pindahkan baris data dengan value `is_valid` NOT VALID dengan kolom `measurement_id, pm25_ug_m3, pm10_ug_m3, so2_ug_m3, co_ug_m3, o3_ug_m3, no2_ug_m3, temperature_c, humidity_pct, wind_speed_ms, wind_direction_deg, dan qc_flag` dengan variable *unvalidate_meas*
3. Ganti value kolom parameter pengukuran (NOT VALID) dengan NaN untuk nantinya di convert ke *float*


```python
# remove miliseconds
def remove_milliseconds(series):
    return series.astype(str).str.split('.').str[0]

# execute
df['timestamp_start'] = remove_milliseconds(df['timestamp_start'])
df['timestamp_end'] = remove_milliseconds(df['timestamp_end'])

# convert to date
df['timestamp_start'] = pd.to_datetime(df['timestamp_start'], format='%Y-%m-%d %H:%M:%S')
df['timestamp_end'] = pd.to_datetime(df['timestamp_end'], format='%Y-%m-%d %H:%M:%S')
```

```python
# qc_flag value
df['qc_flag'].value_counts()
```

```python
# add 'is_valid' coloms
df['is_valid'] = df['qc_flag'].apply(lambda x: x if x == 'VALID' else 'NOT VALID')
df.sample(7, random_state=45)
```

```python
# make unvalidate table for documentation raw data
cols_to_keep = [
    'measurement_id', 'pm25_ug_m3', 'pm10_ug_m3', 'so2_ug_m3', 'co_ug_m3', 
    'o3_ug_m3', 'no2_ug_m3', 'temperature_c', 'humidity_pct', 
    'wind_speed_ms', 'wind_direction_deg', 'qc_flag'
]

aqms_error = df[df['is_valid'] == 'NOT VALID'][cols_to_keep].copy()

aqms_error.sample(7)
```

```python
# cleaning parameter data, fill NaN if is_valid is 'NOT VALID'
cols_to_clean = [
    'pm25_ug_m3', 'pm10_ug_m3', 'so2_ug_m3', 'co_ug_m3', 'o3_ug_m3', 
    'no2_ug_m3', 'temperature_c', 'humidity_pct', 'wind_speed_ms', 
    'wind_direction_deg'
]

df.loc[df['is_valid'] == 'NOT VALID', cols_to_clean] = np.nan

# convert data type to float
for col in cols_to_clean:
    df[col] = df[col].astype(float)

df = df.drop(columns=['qc_flag'], errors='ignore')

# dtypes
print(df.dtypes)
```
